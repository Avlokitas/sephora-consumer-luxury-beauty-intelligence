```python
import ast
import pandas as pd

def count_ingredient_blocks(x):
    if pd.isna(x):
        return 0
    try:
        parsed = ast.literal_eval(x)
        return len(parsed) if isinstance(parsed, list) else 1
    except:
        return 1

df_clean['ingredient_blocks'] = (
    df_clean['ingredients'].apply(count_ingredient_blocks)
)

df_clean['ingredient_blocks'].value_counts().head(10)
```

```python
df_clean['ingredient_blocks'].describe()
```

```python
ingredient_keywords = {
    'Hyaluronic Acid': r'hyaluronic',
    'Niacinamide': r'niacinamide',
    'Vitamin C': r'vitamin c|ascorbic',
    'Retinol': r'\bretinol\b',
    'Salicylic Acid': r'salicylic',
    'Ceramides': r'ceramide',
    'Peptides': r'peptide',
    'Squalane': r'squalane',
    'Glycolic Acid': r'glycolic',
    'Vitamin E': r'tocopherol|vitamin e'
}

for ingredient, pattern in ingredient_keywords.items():
    df_clean[ingredient] = (
        df_clean['ingredients']
        .fillna('')
        .str.contains(pattern, case=False, regex=True)
    )
```

```python
ingredient_summary = []

for ingredient in ingredient_keywords:
    summary = (
        df_clean.groupby('consumer_segment')[ingredient]
        .mean()
        .mul(100)
        .round(2)
        .rename(ingredient)
    )
    ingredient_summary.append(summary)

ingredient_summary = pd.concat(
    ingredient_summary,
    axis=1
).T

ingredient_summary
```

```python
opportunity_categories = (
    df_clean[
        df_clean['consumer_segment'].isin([
            'High Satisfaction, Lower Popularity',
            'Popular, Lower Satisfaction'
        ])
    ]
    .groupby(['primary_category', 'consumer_segment'])
    .agg(
        products=('product_id', 'count'),
        median_loves=('loves_count', 'median'),
        median_rating=('rating', 'median')
    )
    .reset_index()
)

category_totals = (
    df_clean
    .groupby('primary_category')
    .size()
    .rename('category_total')
    .reset_index()
)

opportunity_categories = opportunity_categories.merge(
    category_totals,
    on='primary_category'
)

opportunity_categories['category_share'] = (
    opportunity_categories['products']
    / opportunity_categories['category_total']
    * 100
)

opportunity_categories.round(2)
```

```python
 priority = (
    df_clean
    .groupby(['primary_category', 'consumer_segment'])
    .size()
    .unstack(fill_value=0)
)

priority['discovery_share'] = (
    priority['High Satisfaction, Lower Popularity']
    / priority.sum(axis=1) * 100
)

priority['improvement_share'] = (
    priority['Popular, Lower Satisfaction']
    / priority.sum(axis=1) * 100
)

priority[
    [
        'High Satisfaction, Lower Popularity',
        'Popular, Lower Satisfaction',
        'discovery_share',
        'improvement_share'
    ]
].sort_values(
    'High Satisfaction, Lower Popularity',
    ascending=False
).round(2)
```

```python
discovery_candidates = (
    df_clean[
        (df_clean['consumer_segment'] == 'High Satisfaction, Lower Popularity') &
        (df_clean['primary_category'].isin(['Hair', 'Fragrance', 'Skincare']))
    ]
    .copy()
)

discovery_candidates['discovery_score'] = (
    discovery_candidates['rating'].rank(pct=True) * 0.4
    + discovery_candidates['reviews'].rank(pct=True) * 0.3
    + discovery_candidates['loves_count'].rank(pct=True) * 0.3
)

discovery_candidates = discovery_candidates.sort_values(
    'discovery_score',
    ascending=False
)

discovery_candidates[
    [
        'product_name',
        'brand_name',
        'primary_category',
        'price_usd',
        'loves_count',
        'reviews',
        'rating',
        'child_count',
        'sephora_exclusive',
        'discovery_score'
    ]
].head(30)
```

```python
improvement_candidates = (
    df_clean[
        (df_clean['consumer_segment'] == 'Popular, Lower Satisfaction') &
        (df_clean['primary_category'].isin([
            'Makeup', 'Skincare'
        ]))
    ]
    .copy()
)

improvement_candidates['improvement_score'] = (
    improvement_candidates['loves_count'].rank(pct=True) * 0.5
    + (1 - improvement_candidates['rating'].rank(pct=True)) * 0.5
)

improvement_candidates = improvement_candidates.sort_values(
    'improvement_score',
    ascending=False
)

improvement_candidates[
    [
        'product_name',
        'brand_name',
        'primary_category',
        'price_usd',
        'loves_count',
        'reviews',
        'rating',
        'child_count',
        'sephora_exclusive',
        'improvement_score'
    ]
].head(30)
```
